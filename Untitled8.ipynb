```python
import numpy as np

# Training dataset (Added based on standard ML examples matching your logic)
data = np.array([
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
])

num_attr = len(data[0]) - 1

# Initialize Specific (S) and General (G) boundaries
S = ['0'] * num_attr
G = [['?' for _ in range(num_attr)]]

print(f"Initial Specific Hypothesis S: {S}")
print(f"Initial General Hypothesis G: {G}")

for example in data:
    attributes = example[:-1]
    label = example[-1]

    if label == "Yes":  # Positive Example
        for i in range(num_attr):
            if S[i] == '0':
                S[i] = attributes[i]
            elif S[i] != attributes[i]:
                S[i] = '?'

        # Remove inconsistent hypotheses from G
        G = [g for g in G if all(g[i] == '?' or g[i] == S[i] for i in range(num_attr))]

    else:  # Negative Example
        new_G = []
        for g in G:
            for i in range(num_attr):
                if g[i] == '?':
                    # Create a copy and specialize
                    new_hypothesis = list(g)
                    new_hypothesis[i] = S[i]
                    if new_hypothesis[i] != attributes[i] and S[i] != '?':
                        new_G.append(new_hypothesis)
        G = new_G

    print(f"After Example {example}:")
    print(f"S = {S}")
    print(f"G = {G}")

print(f"\nFinal Specific Boundary: {S}")
print(f"Final General Boundary: {G}")
```

```python
import pandas as pd
import math

# Dataset provided in the file
data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain', 'Rain', 'Overcast', 'Sunny'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal', 'Normal', 'Normal', 'Normal'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak'],
    'PlayTennis': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'Yes']
}
df = pd.DataFrame(data)

def entropy(target):
    values = target.value_counts()
    total = len(target)
    ent = 0
    for v in values:
        p = v / total
        ent -= p * math.log2(p)
    return ent

def info_gain(df, attr, target_name):
    total_entropy = entropy(df[target_name])
    values = df[attr].unique()
    weighted_entropy = 0
    for v in values:
        subset = df[df[attr] == v]
        weighted_entropy += (len(subset) / len(df)) * entropy(subset[target_name])
    return total_entropy - weighted_entropy

def id3(df, target, attributes):
    if len(df[target].unique()) == 1:
        return df[target].iloc[0]

    if not attributes:
        return df[target].mode()[0]

    gains = [info_gain(df, attr, target) for attr in attributes]
    best_attr = attributes[gains.index(max(gains))]

    tree = {best_attr: {}}
    remaining_attrs = [a for a in attributes if a != best_attr]

    for value in df[best_attr].unique():
        subset = df[df[best_attr] == value]
        tree[best_attr][value] = id3(subset, target, remaining_attrs)

    return tree

attributes = list(df.columns[:-1])
tree = id3(df, 'PlayTennis', attributes)
print("Decision Tree:")
print(tree)
```
